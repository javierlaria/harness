```python

def connection_test() -> None:
    provider = os.getenv("PROVIDER", "openai")
    client, model = create_client(provider)
    print("Provider: ", provider)
    print("Client: ", client)
    print("Model: ", model)

    try:
        print("Trying...")
        response = client.chat.completions.create(
            model = model,
            messages =[
                {"role": "user", "content": "Hola, prueba de conexión exitosa."}
            ],
            extra_headers={
                "HTTP-Referer": "http://localhost:3000", 
                "X-Title": "My Local Test Agent",
            }
        )
        print("\nResponse:")
        print(response.choices[0].message.content)
    except OpenAIError as error:
        print("\nAPI request failed: ", error)
    except (KeyError, ValueError) as error:
        print("Didnt work, error:")
        print(error)
```

```python
connection_test()
```

```python
def normalize(text: str) -> str:
    return " ".join(text.lower().strip().split())

def wrap(text: str, wrapper: dict):
    return "".join([wrapper["in"], text, wrapper["out"]])

def exact_match(response: str, expected: str, wrapper: dict) -> bool:
    if wrapper:
        expected = wrap(expected)
    return response.strip().lower() == expected.strip().lower()

def normalized_matching(response: str, expected: str, wrapper: dict) -> float:
    if wrapper:
        expected = wrap(expected)
    return normalize(response) == normalize(expected)

def keywords(response: str, keywords: list[str]) -> bool:
    return all(keyword.lower() in response.lower() for keyword in keywords)
```
